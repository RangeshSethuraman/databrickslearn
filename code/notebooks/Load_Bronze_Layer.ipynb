Loading bronze layer 

```python
# Read nyc trip data from Parquet file into a DataFrame
df = spark.read.parquet("/Volumes/demo/source/inputfiles/nyc_taxi/yellow_tripdata_2026*.parquet")

display(df)
```

```python
# load the data into a table in bronze layer
spark.sql("DROP TABLE IF EXISTS demo.bronze.nyc_taxi")
df.write.saveAsTable("demo.bronze.nyc_taxi")
```

```python
# Read the aqi data into dataframe
df1 = spark.read.csv("/Volumes/demo/source/inputfiles/nyc_aqi/data.csv", header=True , inferSchema=True)
```

```python
#Renaming the df column names to adhear to the bronze schema
df1 = df1.withColumnRenamed("AQI (US)", "AQI") \
.withColumnRenamed("Temperature (°C)", "Temperature") \
.withColumnRenamed("Pressure (hPa)", "Pressure") \
.withColumnRenamed("Humidity (%)", "Humidity") \
.withColumnRenamed("Wind Speed (m/s)", "Wind_Speed") \
.withColumnRenamed("Wind Direction (°)", "Wind_Direction") 
```

```python
# load Air Quality Index data into bronze layer as a table
spark.sql("DROP TABLE IF EXISTS demo.bronze.nyc_aqi")
df1.write.saveAsTable("demo.bronze.nyc_aqi")
```
