```python
# Read data from bronze layer 
from pyspark.sql import functions as F
df = spark.read.table("demo.bronze.nyc_aqi")

# Trim String fileds 
for field in df.schema.fields:
        if isinstance(field.dataType, F.StringType):
            df = df.withColumn(field.name, F.trim(F.col(field.name)))

display(df)


  
  


```

```python
# Insert data into silver layer
df.createOrReplaceTempView("TmpTbl_nyc_aqi")

spark.sql("""
INSERT INTO demo.silver.nyc_aqi
SELECT A.* FROM TmpTbl_nyc_aqi A 
left join demo.silver.nyc_aqi B 
on A.Date = B.Date and A.Time = B.Time and A.City = B.City and A.State = B.State and A.Country = B.Country
where B.date is null
""")
```
